## Mann-Whitney U Test

```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('diabetes.csv')
df.head()
```

### Create two independent gropus

```python
group1 = df[df['Outcome']==0]['Glucose']   # Non-diabetic
group2 = df[df['Outcome']==1]['Glucose']   # diabetic
```

### Perform Mann-Whitney U Test

```python
u_stastic, p_value = stats.mannwhitneyu(group1, group2)
print("U Stastic:", u_stastic)
print("P-value:",p_value)
alpha = 0.05
if p_value < alpha:
    print('Reject H0')
else:
    print("Do not Rejet H0")
```
