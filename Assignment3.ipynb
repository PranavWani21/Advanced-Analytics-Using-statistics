```python
import scipy.stats as stats
from scipy.stats import zscore
```

```python
df = pd.read_csv('diabetes.csv')
df.head()
```

```python
glucose_mean=df['Glucose'].mean()
glucose_std = df['Glucose'].std()
print(f'Mean of Glucose {glucose_mean}')
print(f'Standard deviation of Glucose {glucose_std}')
```

```python
# 1. What is the probability that a randomly selected patient has a glucose level between 100 mg/dL and 140 mg/dL?
prob_100_140_s = stats.norm.cdf(140, loc=glucose_mean, scale=glucose_std) - stats.norm.cdf(100, loc=glucose_mean, scale=glucose_std)
print(f"1. Probability (100 to 140 mg/dL): {prob_100_140_s:.2f}")
```

```python
# Q2: P(X > 180)
# ( 180 - glucose_mean ) / glucose_std = 1.8486277348093851
# 0.5 - 0.4671 = 0.032899999999999985
prob_gt_180_s = 1 - stats.norm.cdf(180, glucose_mean, glucose_std)
print(f"2. Percentage (> 180 mg/dL): {prob_gt_180_s}\n")
```

```python
# 3. Determine the glucose level that separates the top 10% of patients from the remaining 90%
glucose_90th = stats.norm.ppf(0.90,loc=glucose_mean,scale=glucose_std)
print("\n 3. Glucose level for top 10%(90th percentile)")
print(f'{glucose_90th:.2f} mg/dl')
```

```python
# 4. What is the probability that a patient's glucose level lies within one standard deviation of the mean glucose level?
lower = glucose_mean - glucose_std
upper = glucose_mean + glucose_std
prob_within_1sd = (stats.norm.cdf( (upper - glucose_mean)/glucose_std) - stats.norm.cdf((lower - glucose_mean)/glucose_std)  )
print("\n4. Probability within one standard deviation")
print(f'{prob_within_1sd: .4f} ({prob_within_1sd*100:.2f})' )
```

```python
# 5. A patient has a glucose level of 200 mg/dL. Calculate the z-score and determine the probability that a 
#   randomly selected patient has a glucose level higher than this.
z = (200 - glucose_mean) / glucose_std 
prob = 1 - stats.norm.cdf(z)
print("\n5. Patient Glucose = 200")
print(f"zscore = {z}")
print(f'Probability that Glucose > 200 = {prob:.4f} ({prob*100:.2f}%)')
```

```python
# 6 A hospital can only perform expensive follow-up tests on the highest 5% of glucose readings. 
# What glucose level should be used as the cutoff?
glucose_95th = stats.norm.ppf(0.95,loc=glucose_mean,scale=glucose_std)
print("\n 6. Glucose level for top 5%(95th percentile)")
print(f'cutoff glucose level: {glucose_95th:.2f} mg/dl')
```

```python
# 7. What percentage of patients are expected to have glucose levels between the 25th percentile and the 75th percentile?
glucose_25th = stats.norm.ppf(0.25,loc=glucose_mean,scale=glucose_std)
glucose_75th = stats.norm.ppf(0.75,loc=glucose_mean,scale=glucose_std)
print(glucose_25th,glucose_75th)
prob_bet_25th_and_75th = stats.norm.cdf(glucose_75th,loc=glucose_mean,scale=glucose_std) - stats.norm.cdf(glucose_25th,loc=glucose_mean,scale=glucose_std)
print("\nProbability between 25th and 75th percentile: ",prob_bet_25th_and_75th)
print(f"percentage: {prob_bet_25th_and_75th:.4f}")
```

### Q3 Two-Sample t test 

1. Titanic: Is the average age of male passengers significantly different from the average age of female passengers ?

```python
from scipy import stats
```

```python
df = pd.read_csv('titanic.xls')
df.head()
```

```python
df.info()
```

```python
df = df.dropna()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
male = df[df['gender']=="male"]
female = df[df['gender']=="female"]
```

```python
# take the sample
male_sample = male.sample(n=30,replace=True)
female_sample = female.sample(n=30,replace=True)
```

```python
# Perform two sample t-test
t_statistic,p_value = stats.ttest_ind(male_sample['age'],female_sample['age'])
# Degree of Freedom
dof = len(male_sample) + len(female_sample)  - 2

alpha = 0.05 

if p_value < alpha:
    print('There is a significant difference')
else:
    print('There is not significant difference')
```

```python
# Find the critical value
critical_value= stats.t.ppf(1-alpha , dof)

if abs(t_statistic) > critical_value:
    print('There is a significant difference')
else:
    print('there is no significant difference')    
```

```python
# Print the results
print('T-stastic: ', t_statistic)
print('P-value: ',p_value)
print('Critical value: ',critical_value)
```
