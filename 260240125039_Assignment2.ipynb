1. Do an outlier analysis by dsplaying Q1, Q3, IQR, LB, UB, Outlier points and also
   using a box plot on the following dataset
   - kaggle - Billionaires Statistics Dataset

#### Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import  seaborn as sns
```

#### Load Dataset

```python
df=pd.read_csv("Billionaires Statistics Dataset.csv")
df.head()
```

#### Q1, Q3, Lower bound, Upper bound, Outliers

```python
arr=df["finalWorth"]
q1=np.percentile(arr,25)
q3=np.percentile(arr,75)
print("Q1: ",q1)
print("Q3: ",q3)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = []

for x in arr:
    if x < lower_bound or x > upper_bound:
        outliers.append(x)

print("Lower Bound =" ,lower_bound)
print("Upper Bound =" ,upper_bound)
print("outlier Points =" , outliers)
```

#### Boxplot

```python
plt.figure(figsize=(10,6))
sns.boxplot(x=df["finalWorth"])
plt.xlabel("Final Worth")
plt.title("Boxplot")
plt.show()
```

#### Positive Correlation

```python
l1= list(df.columns)
count=0
positive_correlation_columns={}
for i in range(len(l1)):
    for j in range(i+1, len(l1)):
        #print(l1[i],l1[j],end="")
        correlation=df[l1[i]].corr( df[l1[j]])
        if correlation > 0:
            if count==2:
                break
            count+=1
            positive_correlation_columns[count]=[l1[i],l1[j],correlation]
    if count==2:
        break

print(positive_correlation_columns)
```

```python
print(positive_correlation_columns[1])
print(positive_correlation_columns[2])
```

#### Scatter Plot

```python
plt.scatter(df[positive_correlation_columns[1][0]],df[positive_correlation_columns[1][1]])
plt.xlabel(positive_correlation_columns[1][0])
plt.ylabel(positive_correlation_columns[1][1])
plt.title(f'Scatter Plot (Positive Correlation: {round(float(positive_correlation_columns[1][2]),2)})')
plt.legend()
plt.show()
```

3. Use mall-customers.csv dataset
   Perform all the four sampling techniques to pick 10% records. <br>
   For stratified random sampling, consder gender<br>
   For cluster random sampling, consider age brackets (19-29, 30-39, ...)<br>
   Find the average spending score for each of the sampled subsets (x bar). Which one is the closest to the population average(meu)?

```python
df = pd.read_csv("Mall_Customers.csv")
df.head()
```

#### Simple Random Sampling

```python
simple_random_sampling_df = df.sample(frac=0.1,random_state=42)
```

```python
print(len(simple_random_sampling_df))
```

```python
print(f'Average Spending score of original dataset: {np.mean(df['Spending Score (1-100)'])}')
print(f'Average Spending Score of the Sample is: ', np.mean(simple_random_sampling_df['Spending Score (1-100)']))
```

#### Stratified Random Sampling

```python
df['Annual Income (k$)'].min()
```

```python
df['Annual Income (k$)'].max()
```

```python
df_male1 = df[ (df['Gender']=='Male' ) & (df['Annual Income (k$)']<=50) ]
length_df_male1=int(np.ceil(0.1*len(df_male1)))
#print(int(length_df_male1))
df_male_sample1 = df_male1.sample(n=int(length_df_male1),random_state=42)

df_male2 = df[ (df['Gender']=='Male' ) & (df['Annual Income (k$)']>=51) & (df['Annual Income (k$)']<=99) ]
length_df_male2=int(np.ceil(0.1*len(df_male2)))
df_male_sample2 = df_male2.sample(n=length_df_male2,random_state=42)

df_male3 = df[ (df['Gender']=='Male' ) & (df['Annual Income (k$)']>=100) & (df['Annual Income (k$)']<=149) ]
length_df_male3=int(np.ceil(0.1*len(df_male3)))
df_male_sample3 = df_male3.sample(n=length_df_male3,random_state=42)

df_female1 = df[ (df['Gender']=='Female' ) & (df['Annual Income (k$)']<=50) ]
df_female_sample1 = df_female1.sample(n=int(np.ceil(0.1*len(df_female1))),random_state=42)

df_female2 = df[ (df['Gender']=='Female' ) & (df['Annual Income (k$)']>=51) & (df['Annual Income (k$)']<=99) ]
df_female_sample2 = df_female2.sample(n=int(np.ceil(0.1*len(df_female2))),random_state=42)

df_female3 = df[ (df['Gender']=='Female' ) & (df['Annual Income (k$)']>=100) & (df['Annual Income (k$)']<=149) ]
df_female_sample3 = df_female3.sample(n=int(np.ceil(0.1*len(df_female3))),random_state=42)

df_combined = pd.concat([df_male_sample1,df_male_sample2,df_male_sample3,df_female_sample1,df_female_sample2,df_female_sample3])
```

```python
print("Original Dataset Size: ",len(df))
print("Male records with annual income <=50",len(df_male1))
print("Sample Male records with annual income <=50",len(df_male_sample1))

print("\nMale records with annual income >=51 and <=99",len(df_male2))
print("Sample Male records with annual income <=50",len(df_male_sample2))

print("\nMale records with annual income >=99 and <=149",len(df_male3))
print("Sample Male records with annual income <=50",len(df_male_sample3))
```

```python
print("Original Dataset Size: ",len(df))
print("Female records with annual income <=50",len(df_female1))
print("Sample Female records with annual income <=50",len(df_female_sample1))

print("\nFemale records with annual income >=51 and <=99",len(df_female2))
print("Sample Female records with annual income <=50",len(df_female_sample2))

print("\nFemale records with annual income >=99 and <=149",len(df_female3))
print("Sample Female records with annual income <=50",len(df_female_sample3))
```
